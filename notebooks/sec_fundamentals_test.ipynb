# SEC Fundamentals Test Notebook

This notebook tests the SEC fundamentals pipeline for one company and for a portfolio.

The evaluation cells call `data.sec.gov` and cache the raw `companyfacts` response locally. The optional Ollama cell is disabled by default.

```python
from pathlib import Path
import sys

# Make the project root importable whether Jupyter starts here or in notebooks/.
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'core').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

from core.fundamentals import evaluate_portfolio, evaluate_ticker
from core.fundamentals.context import build_fundamental_context

print(f'Project root: {PROJECT_ROOT}')
```

## 1. Evaluate one ticker

Change `TICKER` to any company supported by the SEC ticker mapping. Set `REFRESH = True` when you explicitly want to download the latest company facts again.

```python
TICKER = 'MSFT'
AS_OF = '2026-08-28'
REFRESH = False

company_report = evaluate_ticker(
    TICKER,
    as_of=AS_OF,
    refresh=REFRESH,
)

print(company_report['ticker'], company_report['company_name'])
print('Annual rows:', len(company_report['annual_metrics']))
print('Quarterly rows:', len(company_report['quarterly_metrics']))
print('Quality flags:', company_report['quality_flags'])
```

```python
annual = pd.DataFrame(company_report['annual_metrics'])
quarterly = pd.DataFrame(company_report['quarterly_metrics'])

display(annual.tail(5))
display(quarterly.tail(8))
```

```python
# Inspect the source filing accession numbers attached to the report.
pd.DataFrame(company_report['sources']).tail(10)
```

## 2. Evaluate a portfolio

The portfolio API evaluates each ticker independently and removes duplicate tickers while preserving their first occurrence. Holdings weights are not included yet; this stage is about company fundamentals.

```python
PORTFOLIO = ['AAPL', 'MSFT', 'NVDA']

portfolio_report = evaluate_portfolio(
    PORTFOLIO,
    as_of=AS_OF,
    refresh=REFRESH,
)

portfolio_overview = pd.DataFrame([
    {
        'ticker': report['ticker'],
        'company_name': report['company_name'],
        'annual_rows': len(report['annual_metrics']),
        'quarterly_rows': len(report['quarterly_metrics']),
        'quality_flag_count': len(report['quality_flags']),
    }
    for report in portfolio_report['companies']
])

display(portfolio_overview)
```

## 3. Build bounded LLM context

This context contains recent normalized metrics, quality flags, and filing sources. It does not send the raw SEC JSON to the model.

```python
company_context = build_fundamental_context(company_report)
print(company_context[:6000])
```
